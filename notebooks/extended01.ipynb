# Extended Data Figure 1

Rebuild the complete figure from archived numerical tables and declared SVG artwork. Run all cells in order. No model training, GPU or external editing is required. Shared plotting code is in `src/mapper/figures/`; exact source hashes are in `data/figure_inputs/manifest.json`.

```python
from pathlib import Path
import sys
import json
import matplotlib
matplotlib.use('Agg')
# Works from the notebook directory or the package root.
package = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src/mapper/figures').is_dir())
sys.path.insert(0, str(package / 'src'))
from mapper.figures.registry import build_figure
from mapper.figures.style import sha
manifest = json.loads((package / 'data/figure_inputs/manifest.json').read_text())
assert all(sha(package / 'data/figure_inputs' / name) == record['sha256'] for name, record in manifest['files'].items())
figure_id = 'extended01'
```

## Generate and display

PDF and SVG retain vector artwork and data layers. PNG is exported at 600 dpi. The inline preview is lower-resolution for manageable notebook size.

```python
pdf = build_figure(figure_id)
from IPython.display import Image, display
display(Image(filename=str(pdf.with_name(figure_id + '_preview.png'))))
print('Final files:', pdf.parent)
for suffix in ['.pdf', '.svg', '.png']:
    print(pdf.with_suffix(suffix).name)
```

No prediction scores, significance tests or uncertainty cutoffs are refitted by this notebook. The archived input manifest identifies the analyses that produced them.
